```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
prices=pd.read_csv("Datasets/TUTORIAL_ROUND_1 (1)/prices_round_0_day_-1.csv",sep=";")
trades=pd.read_csv("Datasets/TUTORIAL_ROUND_1 (1)/trades_round_0_day_-1.csv",sep=";")
```

```python
print(prices.columns.tolist())
```

```python
products=prices["product"].unique()
for p in products:
    df=prices[prices["product"]==p]
    plt.plot(df["timestamp"],df["mid_price"],label=p)

plt.legend()
plt.show()
```

```python
for p in trades["symbol"].unique():
    df=trades[trades["symbol"]==p]
    plt.plot(df["timestamp"],df["price"],label=p)

plt.legend()
plt.show()
```
