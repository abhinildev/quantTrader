```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
p1=pd.read_csv("../Datasets/ROUND_1/prices_round_1_day_-1.csv",sep=";")
p2=pd.read_csv("../Datasets/ROUND_1/prices_round_1_day_-2.csv",sep=";")
p0=pd.read_csv("../Datasets/ROUND_1/prices_round_1_day_0.csv",sep=";")
```

```python
t1=pd.read_csv("../Datasets/ROUND_1/trades_round_1_day_-1.csv",sep=";")
t2=pd.read_csv("../Datasets/ROUND_1/trades_round_1_day_-2.csv",sep=";")
t0=pd.read_csv("../Datasets/ROUND_1/trades_round_1_day_0.csv",sep=";")
```

```python
print(p1.columns.tolist())
print(t1.columns.tolist())
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(3, 1, figsize=(10, 12))

# ✅ Define fixed colors
colors = {
    "INTARIAN_PEPPER_ROOT": "blue",
    "ASH_COATED_OSMIUM": "orange"
}

# -------- Day 0 --------
for p in p0["product"].unique():
    df = p0[p0["product"] == p]
    df = df[df["mid_price"].notna()]
    df = df[df["mid_price"] > 0]
    df = df.sort_values("timestamp")

    axs[2].plot(df["timestamp"], df["mid_price"], label=p, color=colors[p])

axs[2].set_title("Day 0")
axs[2].legend()

# -------- Day 1 --------
for p in p1["product"].unique():
    df = p1[p1["product"] == p]
    df = df[df["mid_price"].notna()]
    df = df[df["mid_price"] > 0]
    df = df.sort_values("timestamp")

    axs[0].plot(df["timestamp"], df["mid_price"], label=p, color=colors[p])

axs[0].set_title("Day 1")
axs[0].legend()

# -------- Day 2 --------
for p in p2["product"].unique():
    df = p2[p2["product"] == p]
    df = df[df["mid_price"].notna()]
    df = df[df["mid_price"] > 0]
    df = df.sort_values("timestamp")

    axs[1].plot(df["timestamp"], df["mid_price"], label=p, color=colors[p])

axs[1].set_title("Day 2")
axs[1].legend()

plt.tight_layout()
plt.show()
```
